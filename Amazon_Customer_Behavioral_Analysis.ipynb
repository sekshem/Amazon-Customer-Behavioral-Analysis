```python
freq_map = {
    "Less than once a month": 1,
    "Once a month": 2,
    "Few times a month": 3,
    "Once a week": 4,
    "Multiple times a week": 5
}

df["Frequency_Score"] = df[freq_col].map(freq_map)

print(df["Frequency_Score"].head())
```

```python
age_median = df[age_col].median()

freq_median = df["Frequency_Score"].median()


def segment_customer(row):

    if row[age_col] < age_median and row["Frequency_Score"] >= freq_median:
        return "Young Frequent Buyer"

    elif row[age_col] >= age_median and row["Frequency_Score"] >= freq_median:
        return "Mature Frequent Buyer"

    elif row[age_col] < age_median and row["Frequency_Score"] < freq_median:
        return "Young Occasional Buyer"

    else:
        return "Mature Occasional Buyer"


df["Segment"] = df.apply(segment_customer, axis=1)

print(df["Segment"].value_counts())
```

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=df,
    x=age_col,
    y="Frequency_Score",
    hue="Segment",
    s=100
)

plt.axvline(age_median, linestyle="--")

plt.axhline(freq_median, linestyle="--")

plt.title("Customer Segmentation")

plt.xlabel("Age")

plt.ylabel("Purchase Frequency Score")

plt.tight_layout()

plt.savefig("plots/07_customer_segmentation.png")

plt.show()
```

```python
personas = df.groupby("Segment").agg({
    age_col: "mean",
    "Frequency_Score": "mean",
    category_col: lambda x: x.mode()[0]
})

print(personas)
```
